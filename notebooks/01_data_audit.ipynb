Notebook 01 focused on exploratory data auditing and dataset understanding. The objective was to analyze the dataset structure, identify the target variable, inspect missing values, detect duplicate records, understand feature types, and assess overall data quality before proceeding to preprocessing and feature engineering.

````python
# ==========================================
# MULE ACCOUNT DETECTION
# PHASE 1 : DATA AUDIT
# ==========================================

import pandas as pd
import numpy as np
import os

# ------------------------------------------
# CHECK CURRENT DIRECTORY
# ------------------------------------------

print("=" * 60)
print("CURRENT WORKING DIRECTORY")
print("=" * 60)

print(os.getcwd())

# ------------------------------------------
# LOAD DATASET
# ------------------------------------------

df = pd.read_csv("../data/raw/DataSet.csv")

print("\n" + "=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

# ------------------------------------------
# DATASET SHAPE
# ------------------------------------------

print("\nDATASET SHAPE")
print("-" * 40)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# ------------------------------------------
# FIRST 5 ROWS
# ------------------------------------------

print("\nFIRST 5 ROWS")
print("-" * 40)

print(df.head())

# ------------------------------------------
# DATASET INFO
# ------------------------------------------

print("\nDATASET INFO")
print("-" * 40)

df.info()

# ------------------------------------------
# DATA TYPES COUNT
# ------------------------------------------

print("\nDATA TYPES COUNT")
print("-" * 40)

print(df.dtypes.value_counts())

# ------------------------------------------
# NUMERICAL FEATURES
# ------------------------------------------

numerical_cols = df.select_dtypes(
    include=["int64", "float64"]
).columns

print("\nNUMERICAL FEATURES")
print("-" * 40)

print("Total Numerical Features :", len(numerical_cols))

# ------------------------------------------
# CATEGORICAL FEATURES
# ------------------------------------------

categorical_cols = df.select_dtypes(
    include=["object"]
).columns

print("\nCATEGORICAL FEATURES")
print("-" * 40)

print("Total Categorical Features :", len(categorical_cols))

print("\nCategorical Columns:")

for col in categorical_cols:
    print(col)

# ------------------------------------------
# MISSING VALUE ANALYSIS
# ------------------------------------------

print("\nMISSING VALUE ANALYSIS")
print("-" * 40)

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage":
        (df.isnull().sum().values / len(df)) * 100
})

missing_report = missing_report.sort_values(
    by="Missing_Percentage",
    ascending=False
)

print(missing_report.head(20))

# ------------------------------------------
# COLUMNS > 80% MISSING
# ------------------------------------------

high_missing = missing_report[
    missing_report["Missing_Percentage"] > 80
]

print("\nCOLUMNS WITH >80% MISSING VALUES")
print("-" * 40)

print("Count :", len(high_missing))

print(high_missing.head(20))

# ------------------------------------------
# DUPLICATE RECORDS
# ------------------------------------------

duplicates = df.duplicated().sum()

print("\nDUPLICATE RECORDS")
print("-" * 40)

print("Duplicate Rows :", duplicates)

# ------------------------------------------
# TARGET ANALYSIS
# ------------------------------------------

target_col = "F3924"

print("\nTARGET COLUMN")
print("-" * 40)

print("Target :", target_col)

print("\nTARGET VALUE COUNTS")
print("-" * 40)

print(df[target_col].value_counts())

print("\nTARGET DISTRIBUTION (%)")
print("-" * 40)

print(
    round(
        df[target_col].value_counts(normalize=True) * 100,
        2
    )
)

# ------------------------------------------
# NUMERICAL SUMMARY
# ------------------------------------------

print("\nNUMERICAL SUMMARY")
print("-" * 40)

print(df.describe())

# ------------------------------------------
# CATEGORICAL FEATURE DETAILS
# ------------------------------------------

print("\nCATEGORICAL FEATURE DETAILS")
print("-" * 40)

for col in categorical_cols:

    print(f"\nColumn : {col}")

    print(
        "Unique Values :",
        df[col].nunique()
    )

    print(
        df[col]
        .value_counts()
        .head()
    )

# ------------------------------------------
# SAVE REPORT
# ------------------------------------------

os.makedirs("../reports", exist_ok=True)

missing_report.to_csv(
    "../reports/missing_value_report.csv",
    index=False
)

print("\nMissing Value Report Saved")

# ------------------------------------------
# FINAL SUMMARY
# ------------------------------------------

print("\n" + "=" * 60)
print("FINAL DATASET SUMMARY")
print("=" * 60)

print("Rows                     :", df.shape[0])
print("Columns                  :", df.shape[1])
print("Numerical Features       :", len(numerical_cols))
print("Categorical Features     :", len(categorical_cols))
print("Duplicate Rows           :", duplicates)
print("Target Column            :", target_col)
print("Columns >80% Missing     :", len(high_missing))

print("=" * 60)
````
